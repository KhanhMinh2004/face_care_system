```python
from PIL import Image
import os
from collections import Counter

img_dir = '/content/dataset/train/images'
buckets = Counter()

for f in os.listdir(img_dir):
    if not f.endswith(('.jpg', '.jpeg', '.png')):
        continue
    w, h = Image.open(os.path.join(img_dir, f)).size
    short = min(w, h)
    if short < 400:
        buckets['<400px'] += 1
    elif short < 640:
        buckets['400-640px'] += 1
    elif short < 900:
        buckets['640-900px'] += 1
    else:
        buckets['>900px'] += 1

for k, v in sorted(buckets.items()):
    print(f"{k}: {v} ảnh")
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import os
from PIL import Image

def visualize_resize_comparison(img_dir, label_dir, target=640, num_samples=5):
    """
    Hiển thị ảnh gốc vs ảnh resize để kiểm tra mất đặc trưng
    Vẽ thêm bounding box để thấy rõ mụn trước/sau resize
    """
    # Lấy ảnh nhỏ hơn 400px để test (nhóm quan trọng nhất)
    small_imgs = []
    for fname in os.listdir(img_dir):
        if not fname.endswith(('.jpg', '.jpeg', '.png')):
            continue
        w, h = Image.open(os.path.join(img_dir, fname)).size
        if max(w, h) < 400:
            small_imgs.append(fname)

    samples = small_imgs[:num_samples]
    print(f"Tìm thấy {len(small_imgs)} ảnh <400px, hiển thị {len(samples)} ảnh")

    fig, axes = plt.subplots(len(samples), 3, figsize=(18, 5 * len(samples)))
    if len(samples) == 1:
        axes = [axes]

    for idx, fname in enumerate(samples):
        stem = fname.rsplit('.', 1)[0]

        # ── Đọc ảnh gốc ──
        img_orig = cv2.imread(os.path.join(img_dir, fname))
        img_orig_rgb = cv2.cvtColor(img_orig, cv2.COLOR_BGR2RGB)
        h_orig, w_orig = img_orig.shape[:2]

        # ── Resize ──
        scale = target / max(w_orig, h_orig)
        new_w, new_h = int(w_orig * scale), int(h_orig * scale)
        resized = cv2.resize(img_orig, (new_w, new_h), interpolation=cv2.INTER_LANCZOS4)
        canvas = np.full((target, target, 3), 114, dtype=np.uint8)
        pad_x = (target - new_w) // 2
        pad_y = (target - new_h) // 2
        canvas[pad_y:pad_y+new_h, pad_x:pad_x+new_w] = resized
        canvas_rgb = cv2.cvtColor(canvas, cv2.COLOR_BGR2RGB)

        # ── Đọc label và vẽ bbox ──
        label_path = os.path.join(label_dir, stem + '.txt')
        img_orig_bbox = img_orig_rgb.copy()
        canvas_bbox = canvas_rgb.copy()

        COLORS = [
            (255, 0, 0), (0, 255, 0), (0, 0, 255),
            (255, 165, 0), (128, 0, 128), (0, 255, 255)
        ]

        box_info = []
        if os.path.exists(label_path):
            with open(label_path) as f:
                lines = [l.strip() for l in f if l.strip()]

            for line in lines:
                parts = line.split()
                cls = int(parts[0])
                cx, cy, bw, bh = float(parts[1]), float(parts[2]), float(parts[3]), float(parts[4])
                color = COLORS[cls % len(COLORS)]

                # BBox trên ảnh gốc
                x1 = int((cx - bw/2) * w_orig)
                y1 = int((cy - bh/2) * h_orig)
                x2 = int((cx + bw/2) * w_orig)
                y2 = int((cy + bh/2) * h_orig)
                cv2.rectangle(img_orig_bbox, (x1,y1), (x2,y2), color, 2)

                box_w_px = int(bw * w_orig)
                box_h_px = int(bh * h_orig)
                box_info.append(f"cls{cls}: {box_w_px}x{box_h_px}px")

                # BBox trên ảnh resize (tọa độ tương đối giữ nguyên)
                x1r = int((cx - bw/2) * new_w) + pad_x
                y1r = int((cy - bh/2) * new_h) + pad_y
                x2r = int((cx + bw/2) * new_w) + pad_x
                y2r = int((cy + bh/2) * new_h) + pad_y
                cv2.rectangle(canvas_bbox, (x1r,y1r), (x2r,y2r), color, 2)

                box_w_new = int(bw * new_w)
                box_h_new = int(bh * new_h)
                cv2.putText(canvas_bbox, f"c{cls}:{box_w_new}x{box_h_new}px",
                           (x1r, max(y1r-5, 10)),
                           cv2.FONT_HERSHEY_SIMPLEX, 0.5, color, 1)

        # ── Diff map: vùng nào thay đổi nhiều nhất ──
        orig_crop = img_orig_rgb.astype(np.float32)
        resized_crop = cv2.resize(
            canvas_rgb[pad_y:pad_y+new_h, pad_x:pad_x+new_w],
            (w_orig, h_orig)
        ).astype(np.float32)
        diff = np.abs(orig_crop - resized_crop).mean(axis=2)
        diff_norm = (diff / diff.max() * 255).astype(np.uint8)
        diff_colored = cv2.applyColorMap(diff_norm, cv2.COLORMAP_JET)
        diff_colored = cv2.cvtColor(diff_colored, cv2.COLOR_BGR2RGB)

        # ── Plot ──
        ax0, ax1, ax2 = axes[idx]

        ax0.imshow(img_orig_bbox)
        ax0.set_title(
            f"ẢNH GỐC: {w_orig}x{h_orig}px\n"
            f"Boxes: {', '.join(box_info) if box_info else 'không có box'}",
            fontsize=10, color='red'
        )
        ax0.axis('off')

        ax1.imshow(canvas_bbox)
        ax1.set_title(
            f"SAU RESIZE: {target}x{target}px\n"
            f"Scale: {scale:.2f}x | Pad: ({pad_x},{pad_y})",
            fontsize=10, color='green'
        )
        ax1.axis('off')

        ax2.imshow(diff_colored)
        avg_diff = diff.mean()
        max_diff = diff.max()
        quality = "✅ Tốt" if avg_diff < 15 else "⚠️ Trung bình" if avg_diff < 30 else "❌ Mất nhiều"
        ax2.set_title(
            f"DIFF MAP (đỏ = mất nhiều nhất)\n"
            f"Avg diff: {avg_diff:.1f} | Max: {max_diff:.1f} | {quality}",
            fontsize=10
        )
        ax2.axis('off')

        print(f"\n[{fname}] {w_orig}x{h_orig} → {new_w}x{new_h} (scale {scale:.2f}x)")
        print(f"  Boxes gốc: {', '.join(box_info) if box_info else 'trống'}")
        print(f"  Avg pixel diff: {avg_diff:.1f} — {quality}")

    plt.suptitle(
        f"Kiểm tra chất lượng resize {target}x{target}px\n"
        f"Diff map: xanh = ít thay đổi, đỏ = mất nhiều đặc trưng",
        fontsize=13, fontweight='bold', y=1.01
    )
    plt.tight_layout()
    plt.savefig('/content/resize_quality_check.png', dpi=150, bbox_inches='tight')
    plt.show()
    print("\n✅ Đã lưu ảnh tại /content/resize_quality_check.png")

# Chạy
visualize_resize_comparison(
    img_dir='/content/dataset/train/images',
    label_dir='/content/dataset/train/labels',
    target=640,
    num_samples=5
)
```

```python
from PIL import Image
import os

img_dir = '/content/dataset/train/images'
label_dir = '/content/dataset/train/labels'

small_stats = []

for fname in os.listdir(img_dir):
    if not fname.endswith(('.jpg', '.jpeg', '.png')):
        continue

    img = Image.open(os.path.join(img_dir, fname))
    w, h = img.size

    if max(w, h) >= 400:
        continue

    # Đọc label
    label_path = os.path.join(label_dir, fname.rsplit('.', 1)[0] + '.txt')
    if not os.path.exists(label_path):
        continue

    with open(label_path) as f:
        boxes = [l.strip().split() for l in f if l.strip()]

    for box in boxes:
        bw_px = float(box[3]) * w   # chiều rộng mụn tính bằng pixel
        bh_px = float(box[4]) * h
        small_stats.append((fname, bw_px, bh_px))

# Phân tích kích thước mụn thực tế
tiny = sum(1 for _, bw, bh in small_stats if bw < 20 or bh < 20)
print(f"Tổng boxes trong ảnh <400px: {len(small_stats)}")
print(f"Boxes nhỏ hơn 20px: {tiny} ({100*tiny/len(small_stats):.1f}%)")
print(f"Box nhỏ nhất: {min(min(bw,bh) for _,bw,bh in small_stats):.1f}px")
print(f"Box trung bình: {sum(min(bw,bh) for _,bw,bh in small_stats)/len(small_stats):.1f}px")
```

```python
!pip install ultralytics gdown --upgrade --quiet
from ultralytics import YOLO
from google.colab import files, drive
import numpy as np

drive.mount('/content/drive')

model = YOLO('/content/drive/MyDrive/yolov8_highres_project/acne_darkcircle_detector/trained_weights/best_yolo11_acne.pt')


# Chọn ảnh từ máy để test
uploaded = files.upload()

# Lấy tên ảnh vừa upload
import os
image_path = list(uploaded.keys())[0]

# Dự đoán
results = model.predict(source=image_path, save=True, conf=0.158)
print("Ket qua", results)

result = results[0]
boxes = result.boxes.xyxy.cpu().numpy()      # [x1,y1,x2,y2]
confs = result.boxes.conf.cpu().numpy()      # confidence từng box
img_h, img_w = result.orig_shape

# 👉 Lọc box theo confidence (ví dụ >0.5 cho chắc chắn)
valid_indices = confs > 0.158
boxes = boxes[valid_indices]
confs = confs[valid_indices]

lesion_count = len(boxes)

# 🧮 Tính tổng diện tích mụn
total_box_area = 0
for box in boxes:
    x1, y1, x2, y2 = box
    area = (x2 - x1) * (y2 - y1)
    total_box_area += area

# 🧮 Diện tích toàn ảnh (xem như vùng mặt)
face_area = img_w * img_h

# 📏 Extent (phạm vi ảnh hưởng)
extent = total_box_area / face_area

# 📊 Density (mật độ tổn thương)
density = lesion_count / face_area

# ⭐ Severity score (có thể chỉnh trọng số)
severity_score = 0.6 * extent + 0.4 * density

# 🎯 Phân loại mức độ
if severity_score < 0.01:
    severity = "Mild"
elif severity_score < 0.03:
    severity = "Moderate"
else:
    severity = "Severe"

print(f"Số tổn thương (lesions): {lesion_count}")
print(f"Extent (tỉ lệ diện tích mụn): {extent:.4f}")
print(f"Density (mật độ mụn): {density:.6f}")
print(f"Severity ước lượng: {severity}")
```

```python
# Ensure ultralytics is installed before importing
# !pip install ultralytics --upgrade --quiet

# from ultralytics import YOLO
# import numpy as np

from google.colab import files, drive
drive.mount('/content/drive')
uploaded = files.upload()
!unzip -q data.zip -d /content/dataset
# model = YOLO("/content/drive/MyDrive/yolov8_highres_project/acne_darkcircle_detector/trained_weights/best_yolo11_acne_box0.pt")
# model1 = YOLO("/content/drive/MyDrive/yolov8_highres_project/acne_darkcircle_detector/trained_weights/best_yolo11_acne_box7.pt")
# model2 = YOLO("/content/drive/MyDrive/yolov8_highres_project/acne_darkcircle_detector/trained_weights/best_yolo11_acne_box7.5.pt")
# model3 = YOLO("/content/drive/MyDrive/yolov8_highres_project/acne_darkcircle_detector/trained_weights/best_yolo11_acne_box8.pt")

# metrics = model.val(data='/content/dataset/data.yaml', plots=True)
# metrics1 = model1.val(data='/content/dataset/data.yaml', plots=True)
# metrics2 = model2.val(data='/content/dataset/data.yaml', plots=True)
# metrics3 = model3.val(data='/content/dataset/data.yaml', plots=True)

# # Lấy F1–Confidence curve
# print("Yolo11base")
# conf_values = metrics.curves_results[1][0]
# f1_values = metrics.curves_results[1][1]
# print("Yolo11box7")
# conf_values1 = metrics1.curves_results[1][0]
# f1_values1 = metrics1.curves_results[1][1]
# print("Yolo11box7.5")
# conf_values2 = metrics2.curves_results[1][0]
# f1_values2 = metrics2.curves_results[1][1]
# print("Yolo11box8")
# conf_values3 = metrics3.curves_results[1][0]
# f1_values3 = metrics3.curves_results[1][1]

# # Tìm giá trị F1 cao nhất

# print("Yolo11base")
# f1_mean = f1_values.mean(axis=0)
# best_idx = np.argmax(f1_mean)
# print("Yolo11box7")
# f1_mean1 = f1_values1.mean(axis=0)
# best_idx1 = np.argmax(f1_mean1)
# print("Yolo11box7.5")
# f1_mean2 = f1_values2.mean(axis=0)
# best_idx2 = np.argmax(f1_mean2)
# print("Yolo11box8")
# f1_mean3 = f1_values3.mean(axis=0)
# best_idx3 = np.argmax(f1_mean3)

# print("Yolo11base")
# best_conf = conf_values[best_idx]
# best_f1 = f1_mean[best_idx]
# print("Yolo11box7")
# best_conf1 = conf_values1[best_idx1]
# best_f1=f1_mean1[best_idx1]
# print("Yolo11box7.5")
# best_conf2 = conf_values2[best_idx2]
# best_f1=f1_mean2[best_idx2]
# print("Yolo11box8")
# best_conf3 = conf_values3[best_idx3]
# best_f1=f1_mean3[best_idx3]

# print(f"Best Confidence Threshold Yolo11base: {best_conf:.3f}")
# print(f"Best F1 Score Yolo11: {best_f1:.3f}")

# print(f"Best Confidence Threshold Yolo11box7: {best_conf1:.3f}")
# print(f"Best F1 Score Yolo11box7.5: {best_f1:.3f}")

# print(f"Best Confidence Threshold Yolo11box7.5: {best_conf2:.3f}")
# print(f"Best F1 Score Yolo11box8: {best_f1:.3f}")

# print(f"Best Confidence Threshold Yolo11box8: {best_conf3:.3f}")
# print(f"Best F1 Score Yolo11box8: {best_f1:.3f}")
### cach 2
# Ensure ultralytics is installed
!pip install ultralytics --upgrade --quiet

from ultralytics import YOLO
import numpy as np
from google.colab import drive

drive.mount('/content/drive')

DATA_YAML = '/content/dataset/data.yaml'

models = {
    "YOLO11m": "/content/drive/MyDrive/yolov8_highres_project/acne_darkcircle_detector/trained_weights/best_yolo11m_acne.pt",
    "YOLO11_base": "/content/drive/MyDrive/yolov8_highres_project/acne_darkcircle_detector/trained_weights/best_yolo11_acne_box0.pt",
    "YOLO11_box7": "/content/drive/MyDrive/yolov8_highres_project/acne_darkcircle_detector/trained_weights/best_yolo11_acne_box7.pt",
    "YOLO11_box7.5": "/content/drive/MyDrive/yolov8_highres_project/acne_darkcircle_detector/trained_weights/best_yolo11_acne_box7.5.pt",
    "YOLO11_box8": "/content/drive/MyDrive/yolov8_highres_project/acne_darkcircle_detector/trained_weights/best_yolo11_acne_box8.pt",
}

results = {}

for name, weight in models.items():
    print(f"\n===== Evaluating {name} =====")
    model = YOLO(weight)

    metrics = model.val(
        data=DATA_YAML,
        iou=0.75,          # 🔥 IoU cao để phân biệt box
        plots=False,
        save_json=False
    )

    # ---- Extract metrics ----
    map50_95 = metrics.box.map
    map50 = metrics.box.map50
    map75 = metrics.box.map75
    recall = metrics.box.r.mean() # Fixed: Take the mean of the recall array

    # small / medium / large
    map_small = metrics.box.maps[0]

    # ---- F1 acne only ----
    conf = metrics.curves_results[1][0]
    f1 = metrics.curves_results[1][1]
    acne_class_idx = 0   # đổi nếu acne không phải class 0
    f1_acne = f1[:, acne_class_idx]
    best_f1_acne = f1_acne.max()
    best_conf = conf[np.argmax(f1_acne)]

    results[name] = {
        "mAP50": map50,
        "mAP75": map75,
        "mAP50-95": map50_95,
        "Recall@0.75": recall,
        "mAP_small": map_small,
        "Best_F1_acne": best_f1_acne,
        "Best_conf_acne": best_conf
    }

# ===== SUMMARY =====
print("\n\n=========== SUMMARY (Higher is better) ===========")
for name, r in results.items():
    print(f"\n{name}")
    for k, v in r.items():
        print(f"{k:15s}: {v:.4f}")
```

```python
from ultralytics import YOLO
import numpy as np

models = {
    "model_1": "/content/11m_tuned/best.pt",
    "model_2": "/content/11m_default/best.pt",
    "model_3": "/content/11s_tuned/best.pt",
}

results_summary = []

for name, path in models.items():
    model = YOLO(path)

    metrics = model.val(
        data="/content/dataset/data.yaml",
        imgsz=832,
        conf=None
    )

    # ===== Basic metrics =====
    precision = metrics.box.p.mean()
    recall = metrics.box.r.mean()
    map50 = metrics.box.map50
    map50_95 = metrics.box.map

    # ===== F1 từ YOLO =====
    f1_per_class = metrics.box.f1
    f1_mean = np.mean(f1_per_class)

    results_summary.append({
        "model": name,
        "precision": float(precision),
        "recall": float(recall),
        "F1-score": float(f1_mean),
        "mAP50": float(map50),
        "mAP50-95": float(map50_95)
    })

for r in results_summary:
    print(r)
```

```python
from ultralytics import YOLO
import numpy as np

model = YOLO("/content/11m_tuned/best.pt")

conf_range = np.arange(0.001, 0.25, 0.01)

best_conf = 0
best_map = 0

results_conf = []

for conf in conf_range:
    metrics = model.val(
        data="/content/dataset/data.yaml",
        split="test",
        imgsz=832,
        conf=round(conf, 2),
        verbose=False
    )

    map50 = metrics.box.map50

    results_conf.append((round(conf, 2), map50))

    print(f"conf={conf:.2f} → mAP50={map50:.4f}")

    if map50 > best_map:
        best_map = map50
        best_conf = conf

print("\n🔥 BEST CONF:", round(best_conf, 2))
print("🔥 BEST mAP50:", best_map)
```
